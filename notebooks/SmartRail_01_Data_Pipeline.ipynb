```python
import pandas as pd

df = pd.read_csv("MetroPT3(AirCompressor).csv")
df.head()
```

```python
df.info()
```

```python
df.drop(columns=["Unnamed: 0"], inplace=True)
df["timestamp"] = pd.to_datetime(df["timestamp"])
df.describe()
```

```python
df.set_index("timestamp", inplace=True)
df.sort_index(inplace=True)
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
df.shape
```

```python
time_diffs = df.index.to_series().diff()
time_diffs.value_counts().head(10)
```

```python
df.to_pickle("metropt3_clean.pkl")
```

```python
failure_periods = [
    ("2020-04-18 00:00:00", "2020-04-18 23:59:00"),
    ("2020-05-29 23:30:00", "2020-05-30 06:00:00"),
    ("2020-06-05 10:00:00", "2020-06-07 14:30:00"),
    ("2020-07-15 14:30:00", "2020-07-15 19:00:00"),
]

df["failure"] = 0
for start, end in failure_periods:
    df.loc[start:end, "failure"] = 1

df["failure"].value_counts()
```

```python
sensor_cols = ["TP2", "TP3", "H1", "DV_pressure", "Reservoirs",
               "Oil_temperature", "Motor_current"]

df[sensor_cols].plot(subplots=True, figsize=(15, 20), title=sensor_cols)
```

```python
window = 30

feature_df = pd.DataFrame(index=df.index)

for col in sensor_cols:
    feature_df[f"{col}_roll_mean"] = df[col].rolling(window).mean()
    feature_df[f"{col}_roll_std"] = df[col].rolling(window).std()
    feature_df[f"{col}_roll_min"] = df[col].rolling(window).min()
    feature_df[f"{col}_roll_max"] = df[col].rolling(window).max()

feature_df.shape
```

```python
for col in sensor_cols:
    feature_df[f"{col}_diff"] = df[col].diff()
    feature_df[f"{col}_lag1"] = df[col].shift(1)

feature_df["failure"] = df["failure"]
feature_df.dropna(inplace=True)
feature_df.shape
```
